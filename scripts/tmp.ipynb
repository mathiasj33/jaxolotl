```python
from jaxolotl.environments.zone_env.zone_env import ZoneEnv
from jaxolotl.struct_ltl.curriculum.formula_cache import FormulaCache


propositions = ZoneEnv.propositions
assignments = ZoneEnv.assignments()
cache = FormulaCache([p for p in propositions if p != "purple"], assignments)
all_formulas = cache.props
```

```python
import random

from jaxolotl.ltl.logic.assignment import Assignment
from jaxolotl.ltl.logic.boolean_parser import BooleanNode
from jaxolotl.ltl.logic.utils import compute_sat


def sample_reach(depth: int, all_formulas: list[BooleanNode], assignments: list[Assignment]) -> str:
    formulas = [random.choice(all_formulas)]
    for _ in range(depth - 1):
        current_sat = compute_sat(formulas[-1], tuple(assignments))
        while True:
            next_formula = random.choice(all_formulas)
            next_sat = compute_sat(next_formula, tuple(assignments))
            if len(next_sat & current_sat) == 0:
                formulas.append(next_formula)
                break
    formulas = formulas[::-1]
    formula_str = f"F {formulas[0]}"
    for f in formulas[1:]:
        formula_str = f"F ({f} & {formula_str})"
    return formula_str
```

```python
formulas = [sample_reach(2, all_formulas, assignments) for _ in range(3)]
for f in formulas:
    print(f"- \"{f}\"")
```

```python
from jaxolotl.environments.warehouse_env.warehouse_env import WarehouseEnv

propositions = WarehouseEnv.propositions
assignments = WarehouseEnv.assignments()
cache = FormulaCache(propositions, assignments)
all_formulas = cache.props + cache.ands + cache.and_nots
```

```python
formulas = [sample_reach(2, all_formulas, assignments) for _ in range(3)]
for f in formulas:
    print(f"- \"{f}\"")
```
